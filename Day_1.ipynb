```python
def calculate_bill(amount, discount=0, tax=5):
    after_discount = amount - (amount * discount / 100)
    final_amount = after_discount + (after_discount * tax / 100)
    return final_amount

print(calculate_bill(2000))
print(calculate_bill(2000, discount=10))
print(calculate_bill(2000, discount=20, tax=18))
```

```python
with open("student.csv", newline="", encoding="utf-8") as f:
    students=list(csv.DictReader(f))

print("number of students:", len(students))
print(students[0])
```

```python
for s in students:
    s["Roll Number"]=int(s["Roll Number"])
    for subject in ["Maths","Physics","Programming"]:
        s[subject]=int(s[subject])
print(students[0])
```
