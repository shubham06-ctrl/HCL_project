```python
#     Find the number of repeated ID's?

import csv
with open("mobile_phone_sales_data.csv", newline="", encoding="utf-8") as f:
    rows=list(csv.DictReader(f))

repeated = []
one_time = []

for row in rows:
    if row["Sale_ID"] in one_time:
        repeated.append(row["Sale_ID"])
    else:
        one_time.append(row["Sale_ID"])


print("Number of repeated ID's",len(repeated))
print(repeated)
```

```python
#NUMBER OF ROWS LEFT AFTER REMOVING ROWS WHICH HAVE BLANK VALUE OR NULL VALUE 

import pandas as pd

df = pd.read_csv("mobile_phone_sales_data.csv")

new_df = df.dropna()

print("Original DataFrame rows:", len(df))
print("New DataFrame rows:", len(new_df))

difference = len(df) - len(new_df)

print("Difference in rows:", difference)
```

```python
# Fill the empty and null values 


import pandas as pd

df = pd.read_csv('mobile_phone_sales_data.csv')


df.fillna({"City": "Ghaziabad"}, inplace=True)  #Fill with constant 


x = df["Marketing_Spend_INR"].mean()     #Fill with mean
df.fillna({"Marketing_Spend_INR": x}, inplace=True)


x = df["Unit_Price_INR"].mode()[0]     #fill with mode
df.fillna({"Unit_Price_INR": x}, inplace=True)

df.fillna({"Sales_Channel": "Mobile apps"}, inplace=True)


```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("mobile_phone_sales_formatted.csv")

print(df["Net_Sales_INR"].describe())
 
plt.figure(figsize=(8,5))

sns.histplot(df["Net_Sales_INR"], bins=10, kde=True)

plt.title("Distribution of Net Sales")
plt.xlabel("Net Sales (INR)")
plt.ylabel("Number of Sales Records")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("mobile_phone_sales_formatted.csv")

plt.figure(figsize=(8,5))

sns.scatterplot(data=df,
                x="Unit_Price_INR",
                y="Net_Sales_INR")

plt.title("Unit Price vs Net Sales")
plt.xlabel("Unit Price (INR)")
plt.ylabel("Net Sales (INR)")

plt.show()


plt.figure(figsize=(8,5))

sns.scatterplot(data=df,
                x="Units_Sold",
                y="Net_Sales_INR")

plt.title("Units Sold vs Net Sales")
plt.xlabel("Units Sold")
plt.ylabel("Net Sales (INR)")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("mobile_phone_sales_formatted.csv")

# Brand-wise Sales
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Brand",
            y="Net_Sales_INR", estimator="mean")

plt.title("Average Net Sales by Brand")
plt.xticks(rotation=45)
plt.show()


# City-wise Sales
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="City",
            y="Net_Sales_INR", estimator="mean")

plt.title("Average Net Sales by City")
plt.xticks(rotation=45)
plt.show()


# Sales Channel-wise Sales
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Sales_Channel",
            y="Net_Sales_INR", estimator="mean")

plt.title("Average Net Sales by Sales Channel")
plt.show()


# Payment Mode-wise Sales
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Payment_Mode",
            y="Net_Sales_INR", estimator="mean")

plt.title("Average Net Sales by Payment Mode")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("mobile_phone_sales_formatted.csv")

df["Sales_Category"] = df["Net_Sales_INR"].apply(
    lambda x: "Low Revenue" if x < 150000 else "High Revenue"
)

# Filter Low Revenue Records
low_sales = df[df["Net_Sales_INR"] < 150000]

print("Low Revenue Records:")
print(low_sales)

print("Total Low Revenue Records:", len(low_sales))

# Graph
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="Sales_Category")

plt.title("Low Revenue vs High Revenue Sales")
plt.xlabel("Sales Category")
plt.ylabel("Number of Sales Records")

plt.show()
```
