```python
import csv
with open("mobile_phone_sales_data.csv", newline="", encoding="utf-8") as f:
    rows=list(csv.DictReader(f))
    columns=list(rows[0].keys())
print("number of columns:", len(columns))
print("number of rows:", len(rows))
print("column names:", columns)


```

```python
repeated = []
seen = []

for row in rows:
    if row["Sale_ID"] in seen:
        repeated.append(row["Sale_ID"])
    else:
        seen.append(row["Sale_ID"])

print(repeated)
```
